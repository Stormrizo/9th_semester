```python
#!/usr/bin/env python3

import numpy as np
import matplotlib.pyplot as plt

# ==============================
#  Configuración general
# ==============================
plt.rcParams.update({
    "figure.figsize": (6.0, 4.0),   # un poco más compacto
    "font.size": 12,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "grid.alpha": 0.25
})

# ==============================
#  Familia F_{p0}(x) = p0 x - x^2/2
# ==============================
x = np.linspace(-11, 11, 1500)      # rango un poco más corto
p_vals = np.arange(-10, 11, 1)      # -10, -9, ..., 9, 10

fig, ax = plt.subplots()

for p0 in p_vals:
    F = p0 * x - 0.5 * x**2
    # líneas de fondo en gris muy fino
    ax.plot(x, F, linewidth=0.8, color="0.65")

# Ahora dibujamos sólo tres curvas resaltadas
for p0, col in [(-10, "tab:purple"), (0, "tab:blue"), (10, "tab:orange")]:
    F = p0 * x - 0.5 * x**2
    ax.plot(x, F, linewidth=2.0, label=fr"$p_0={p0}$", color=col)

ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$F_{p_0}(x) = p_0 x - \dfrac{x^{2}}{2}$")

# Ejes
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)

# Límites sugeridos (ajusta si quieres otro encuadre)
ax.set_xlim(-11, 11)
ax.set_ylim(-200, 60)

# Leyenda más discreta
ax.legend(loc="lower left", fontsize=9, frameon=False)

fig.tight_layout()

# Guardar con alta calidad
fig.savefig("F_parabolas_p-10_10_mejorado.png", dpi=500, bbox_inches="tight")
fig.savefig("F_parabolas_p-10_10_mejorado.pdf", bbox_inches="tight")

plt.close(fig)
```

```python
#!/usr/bin/env python3

import numpy as np
import matplotlib.pyplot as plt

# ==============================
#  Configuración general
# ==============================
plt.rcParams.update({
    "figure.figsize": (7.0, 4.5),   # figura un poco más grande
    "font.size": 12,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "grid.alpha": 0.25
})

# ==============================
#  Familia F_{p0}(x) = p0 x - x^2/2
# ==============================
x = np.linspace(-11, 11, 1500)
p_vals = np.arange(-10, 11, 1)             # -10, -9, ..., 9, 10
colores = plt.cm.viridis(
    np.linspace(0, 1, len(p_vals))
)                                          # 21 colores distintos

fig, ax = plt.subplots()

# 1) Dibujar TODAS las curvas con color propio
for i, p0 in enumerate(p_vals):
    F = p0 * x - 0.5 * x**2
    ax.plot(x, F, linewidth=1.1, color=colores[i])

# 2) Volver a dibujar SOLO p0=-10,0,10 más gruesas y con leyenda
for p0 in (-10, 0, 10):
    idx = np.where(p_vals == p0)[0][0]     # índice en el arreglo de colores
    F = p0 * x - 0.5 * x**2
    ax.plot(
        x, F,
        linewidth=2.4,
        color=colores[idx],
        label=fr"$p_0={p0}$"
    )

ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$F_{p_0}(x) = p_0 x - \dfrac{x^{2}}{2}$")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)

ax.set_xlim(-11, 11)
ax.set_ylim(-200, 60)

ax.legend(loc="lower left", fontsize=9, frameon=False)

fig.tight_layout()

fig.savefig("F_parabolas_p-10_10_color.png", dpi=500, bbox_inches="tight")
fig.savefig("F_parabolas_p-10_10_color.pdf", bbox_inches="tight")

plt.close(fig)
```

```python
#!/usr/bin/env python3

import numpy as np
import matplotlib.pyplot as plt

# ==============================
#  Configuración general (escala ENORME)
# ==============================
plt.rcParams.update({
    "figure.figsize": (12.0, 7.0),   # tamaño físico grande (pulgadas)
    "font.size": 18,                 # letras grandes
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "grid.alpha": 0.25
})

# ==============================
#  Familia F_{p0}(x) = p0 x - x^2/2
# ==============================
x = np.linspace(-11, 11, 2500)      # más puntos para suavidad extrema
p_vals = np.arange(-10, 11, 1)      # -10, -9, ..., 9, 10
colores = plt.cm.viridis(
    np.linspace(0, 1, len(p_vals))
)

fig, ax = plt.subplots()

# 1) Todas las curvas con su color
for i, p0 in enumerate(p_vals):
    F = p0 * x - 0.5 * x**2
    ax.plot(x, F, linewidth=1.5, color=colores[i])

# 2) Resaltar p0 = -10, 0, 10 con la misma paleta
for p0 in (-10, 0, 10):
    idx = np.where(p_vals == p0)[0][0]
    F = p0 * x - 0.5 * x**2
    ax.plot(
        x, F,
        linewidth=3.0,
        color=colores[idx],
        label=fr"$p_0={p0}$"
    )

ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$F_{p_0}(x) = p_0 x - \dfrac{x^{2}}{2}$")

ax.axhline(0, color="black", linewidth=1.0)
ax.axvline(0, color="black", linewidth=1.0)

ax.set_xlim(-11, 11)
ax.set_ylim(-200, 60)

ax.legend(loc="lower left", fontsize=16, frameon=False)

fig.tight_layout()

# ===== Guardado con resolución bestial =====
# Esto te da ~10800 x 6300 píxeles
fig.savefig("F_parabolas_p-10_10_gigante.png", dpi=900, bbox_inches="tight")
# Y el PDF vectorial para LaTeX (recomendado en el reporte)
fig.savefig("F_parabolas_p-10_10_gigante.pdf", bbox_inches="tight")

plt.close(fig)
```
