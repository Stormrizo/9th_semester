```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos (pueden cambiarse)
eps = 8.854e-12      # permittivity (por ejemplo, vacío)
mu  = 4e-7 * np.pi   # permeability (vacío)
A0  = 1.0            # amplitud del campo eléctrico
lam = 1.0            # longitud de onda
k   = 2.0 * np.pi / lam
omega = 2.0 * np.pi  # frecuencia angular (por ejemplo, 1 Hz en unidades arbitrarias)

# Malla espacial 1D
z_min, z_max = 0.0, 2.0 * lam
z = np.linspace(z_min, z_max, 500)

# Tiempo fijo para "congelar" la onda
t0 = 0.0  # puedes probar con otros valores

# ---------- Objetivo 1: E(z) ----------
Ex = A0 * np.cos(k * z - omega * t0)

plt.figure()
plt.plot(z, Ex)
plt.xlabel("z")
plt.ylabel("E_x(z, t0)")
plt.title("Electric field of a plane wave at t = {:.2f}".format(t0))
plt.grid(True)
plt.tight_layout()
plt.show()

# ---------- Objetivo 2: u(z) ----------
# Para onda plana en medio sin pérdidas: u = eps * |E|^2
u = eps * Ex**2

plt.figure()
plt.plot(z, u)
plt.xlabel("z")
plt.ylabel("u(z, t0)")
plt.title("Electromagnetic energy density at t = {:.2f}".format(t0))
plt.grid(True)
plt.tight_layout()
plt.show()
```
