```python
# Dispersión de electrón libre: epsilon(k) = ħ^2 k^2 / (2m)
import numpy as np
import matplotlib.pyplot as plt

# --- Constantes físicas ---
hbar = 1.054_571_817e-34   # J·s
m_e  = 9.109_383_7015e-31  # kg (masa del electrón)
eV   = 1.602_176_634e-19   # J/eV

# --- Parámetros de la gráfica ---
m      = m_e               # masa efectiva (ajusta si quieres m* != m_e)
kmax   = 1.5e10            # [1/m] rango de k (ajustable)
N      = 2000              # puntos
k      = np.linspace(-kmax, kmax, N)

# --- Dispersión ---
E_J  = (hbar**2 * k**2) / (2*m)        # energía en Joules
E_eV = E_J / eV                        # energía en eV

# --- Gráfica ---
plt.figure(figsize=(7.0,4.2))
plt.plot(k, E_eV, label=r'$\epsilon(k)=\hbar^2 k^2/2m$')
plt.xlabel(r'$k\ \mathrm{[m^{-1}]}$')
plt.ylabel(r'$E\ \mathrm{[eV]}$')
plt.title('Electrón libre: Dispersión parabólica (1D)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# --- (Opcional) Curva sólo para k>=0: E vs |k| ---
kpos = np.linspace(0, kmax, N)
Epos = (hbar**2 * kpos**2) / (2*m) / eV
plt.figure(figsize=(7.0,4.2))
plt.plot(kpos, Epos, label=r'$E(|k|)$')
plt.xlabel(r'$|k|\ \mathrm{[m^{-1}]}$')
plt.ylabel(r'$E\ \mathrm{[eV]}$')
plt.title('Electrón libre: $E$ vs $|k|$')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
