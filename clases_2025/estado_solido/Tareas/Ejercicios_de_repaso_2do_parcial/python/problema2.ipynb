```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros (ajústalos si quieres)
a = 1.0      # parámetro de red
beta = 1.0   # constante del resorte
m = 1.0      # masa atómica

# Rango de k desde 0 hasta el borde de zona (pi/a)
k = np.linspace(0.0, np.pi/a, 400)

# Dispersión (dos ramas idénticas y una plana)
omega = 2.0*np.sqrt(beta/m)*np.abs(np.sin(0.5*k*a))   # ramas 1 y 2
omega_zero = np.zeros_like(k)                         # rama 3 (plana)

# Valor máximo para referencia
omega_max = 2.0*np.sqrt(beta/m)

# Figura
plt.figure(figsize=(7,4.2))
plt.plot(k, omega, label=r'$\omega_1(k)$')
plt.plot(k, omega, label=r'$\omega_2(k)$')  # misma curva (degenerada)
plt.plot(k, omega_zero, label=r'$\omega_3(k)=0$')

# Marcas útiles
plt.axhline(omega_max, linestyle='--', linewidth=1)
plt.text(0.02*(np.pi/a), omega_max*1.02, r'$\omega_{\max}=2\sqrt{\beta/m}$')

# Etiquetas y ejes
plt.xlabel(r'$k$')
plt.ylabel(r'$\omega(k)$')
plt.title(r'Bosquejo de $\omega(k)$: dos ramas degeneradas y una rama plana')
plt.xlim(0, np.pi/a)
plt.ylim(-0.05*omega_max, 1.1*omega_max)
plt.grid(True, alpha=0.3)
plt.legend()

# (Opcional) Colocar ticks simbólicos: Γ en 0 y X en π/a
plt.xticks([0.0, np.pi/(2*a), np.pi/a], [r'$\Gamma$', r'$\cdots$', r'$X$'])

plt.tight_layout()
plt.show()
```

```python
# dispersion_3_ramas.py
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ===== Parámetros (ajústalos si quieres) =====
a = 1.0       # parámetro de red
beta = 1.0    # constante del resorte
m = 1.0       # masa atómica

# ===== Opciones de guardado =====
SAVE_PNG = True
OUT_DIR = Path(".")              # carpeta de salida
FILENAME = "dispersion_3_ramas.png"
DPI = 300                        # resolución del PNG

# ===== Cálculos =====
k = np.linspace(0.0, np.pi/a, 400)  # 0 -> borde de zona (π/a)
omega = 2.0*np.sqrt(beta/m)*np.abs(np.sin(0.5*k*a))   # ramas 1 y 2
omega_zero = np.zeros_like(k)                         # rama 3 (plana)
omega_max = 2.0*np.sqrt(beta/m)

# ===== Gráfica =====
plt.figure(figsize=(7, 4.2))
plt.plot(k, omega, label=r'$\omega_1(k)$')
plt.plot(k, omega, label=r'$\omega_2(k)$')  # misma curva (degenerada)
plt.plot(k, omega_zero, label=r'$\omega_3(k)=0$')

# Línea de referencia en ω_max
plt.axhline(omega_max, linestyle='--', linewidth=1)
plt.text(0.02*(np.pi/a), omega_max*1.02, r'$\omega_{\max}=2\sqrt{\beta/m}$')

# Ejes, título y estilo
plt.xlabel(r'$k$')
plt.ylabel(r'$\omega(k)$')
plt.title(r'Bosquejo de $\omega(k)$: dos ramas degeneradas y una rama plana')
plt.xlim(0, np.pi/a)
plt.ylim(-0.05*omega_max, 1.1*omega_max)
plt.grid(True, alpha=0.3)
plt.legend()

# Ticks simbólicos: Γ y X
plt.xticks([0.0, np.pi/(2*a), np.pi/a], [r'$\Gamma$', r'$\cdots$', r'$X$'])

plt.tight_layout()

# ===== Guardar PNG =====
if SAVE_PNG:
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    out_path = OUT_DIR / FILENAME
    plt.savefig(out_path, dpi=DPI, bbox_inches='tight')
    print(f"Figura guardada en: {out_path.resolve()}")

# Mostrar en pantalla (si quieres, deja esta línea)
plt.show()
```
